2. *Even or Odd*
   Write a program that takes a number from the user and prints whether the number is *Even* or *Odd*.

```python
n=int(input("Enter the number="))
if n % 2 ==0:
    print(f"{n} is Even")
else:
    print(f"{n} is Odd")
```

3. *Grade Calculator*
   Given a student's mark (0 to 100), print the grade based on these rules:
  *90 or above:* Grade A
   *75 to 89:* Grade B
   *50 to 74:* Grade C
    *Below 50:* Fail

```python
mark = int(input("Enter the student mark: "))
if mark >= 90:
    print("Grade A")
elif mark >= 75:
    print("Grade B")
elif mark >= 50:
    print("Grade C")
else:
    print("Fail")
```

4. *Multiplication Table*
   Write a `for` loop that prints the multiplication table for the number **5** (from `5 x 1` up to `5 x 10`).


```python
num = 5
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

5. *Find the Maximum*
   Write a Python function named `find_max(a, b)` that takes two numbers as arguments and returns the larger number.

```python
def find_max(a, b):
    if a > b:
        return a
    else:
        return b

# Demonstrate the function
num1 = 15
num2 = 27
maximum_number = find_max(num1, num2)
print(f"The maximum of {num1} and {num2} is: {maximum_number}")
```
